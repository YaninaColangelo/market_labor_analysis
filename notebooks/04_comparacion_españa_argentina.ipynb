Objetivo: comparar patrones de demanda entre paises, cuidando que la lectura no se limite a frecuencias de herramientas.

```python
from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
df = pd.read_csv(PROJECT_ROOT / 'data' / 'processed' / 'ofertas_clasificadas.csv')
df.head()
```

```python
tabla_pais_necesidad = pd.crosstab(df['pais'], df['necesidad_organizacional'])
tabla_pais_necesidad
```

```python
df.groupby('pais')['nivel_orientacion_negocio'].mean().sort_values(ascending=False)
```
