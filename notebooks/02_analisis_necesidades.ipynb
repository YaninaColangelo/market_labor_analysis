# 02 - Analisis de necesidades organizacionales

Objetivo: analizar que problemas buscan resolver las empresas cuando publican ofertas para Data Analyst.

```python
from pathlib import Path
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
df = pd.read_csv(PROJECT_ROOT / 'data' / 'processed' / 'ofertas_clasificadas.csv')
df.head()
```

```python
necesidades = df['necesidad_organizacional'].value_counts().reset_index()
necesidades.columns = ['necesidad_organizacional', 'cantidad']
necesidades
```

```python
plt.figure(figsize=(10, 4))
sns.barplot(data=necesidades, y='necesidad_organizacional', x='cantidad', color='#2f6f73')
plt.title('Necesidades organizacionales detectadas')
plt.xlabel('Cantidad de ofertas')
plt.ylabel('')
plt.tight_layout()
```
