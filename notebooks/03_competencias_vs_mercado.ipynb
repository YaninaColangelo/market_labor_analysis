# 03 - Competencias vs mercado

Objetivo: observar como se combinan competencias analiticas, de negocio, tecnicas y comunicacionales dentro de la demanda laboral.

```python
from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
df = pd.read_csv(PROJECT_ROOT / 'data' / 'processed' / 'ofertas_clasificadas.csv')
df[['id_oferta', 'competencias_requeridas', 'herramientas_mencionadas']]
```

```python
competencias = (
    df.assign(competencia=df['competencias_requeridas'].str.split('; '))
      .explode('competencia')
      .groupby('competencia', dropna=False)
      .size()
      .sort_values(ascending=False)
)
competencias
```

Nota analitica: interpretar cada competencia segun el valor que permite generar. Por ejemplo, visualizacion no es solo producir graficos; puede representar claridad ejecutiva, seguimiento operativo o coordinacion entre areas.
