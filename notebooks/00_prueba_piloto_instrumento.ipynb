````python
def detectar_herramientas(texto):

    texto = texto.lower()

    herramientas = []

    if "sql" in texto:
        herramientas.append("SQL")

    if "python" in texto:
        herramientas.append("PY")

    # Evitar confundir la letra "r" con cualquier palabra
    if " r " in f" {texto} " or "rstudio" in texto:
        herramientas.append("R")

    if "power bi" in texto:
        herramientas.append("PBI")

    if "excel" in texto:
        herramientas.append("EX")

    if "tableau" in texto:
        herramientas.append("TB")

    return herramientas
````

````python
def detectar_necesidad(texto):

    texto = texto.lower()

    if any(p in texto for p in [
        "indicador",
        "seguimiento",
        "kpi",
        "métrica",
        "dashboard"
    ]):
        return "N02"

    if any(p in texto for p in [
        "integrar",
        "consolidar",
        "distintas fuentes"
    ]):
        return "N03"

    if any(p in texto for p in [
        "automatizar",
        "automatización"
    ]):
        return "N04"

    return None
````

````python
validacion = pd.DataFrame()

validacion["id_oferta"] = df["id_oferta"]

validacion["herramientas"] = (
    df["requisitos"]
    .apply(detectar_herramientas)
)

validacion["necesidad_principal"] = (
    df["descripcion"] + " " + df["responsabilidades"]
).apply(detectar_necesidad)

validacion
````
