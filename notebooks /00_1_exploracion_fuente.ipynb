# **Extractor**

# 1.1 — Conectividad

```python
url = "https://www.tecnoempleo.com/analista-datos-powerbi-sopra-steria/power-bi-etl/rf-7624176cc2dbe32ca545"

headers = {"User-Agent": "Mozilla/5.0"}

response = requests.get(url, headers=headers)

print(response.status_code)
```

# 1.2 — Crear el objeto BeautifulSoup

```python
soup = BeautifulSoup(response.text, "html.parser")
```

# 1.3 — Primera extracción

```python
meta_title = soup.find("meta", property="og:title")

print(meta_title)
```

```python
meta_description = soup.find("meta",attrs={"name": "description"})

print(meta_description)
```

# 1.4 — Validación

```python
descripcion_meta = meta_description.get("content")

print(descripcion_meta)
```

# 1.5 — Construir el primer diccionario

```python
titulo = meta_title.get("content") if meta_title else None
descripcion = meta_description.get("content") if meta_description else None

oferta = {
    "id_oferta": None,
    "fecha_publicacion": None,
    "fuente": "Tecnoempleo",
    "url": url,
    "pais": "España",
    "ciudad": None,
    "empresa": None,
    "industria": None,
    "titulo_puesto": titulo,
    "descripcion": descripcion,
    "responsabilidades": None,
    "requisitos": None,
    "modalidad": None,
    "seniority": None,
    "formacion_requerida": None
}
```

```python
oferta
```

# **Transformador**

# Primera transformación

```python
def separar_titulo_ciudad(texto):

    if " - " in texto:
        titulo, ciudad = texto.rsplit(" - ", 1)
        return titulo.strip(), ciudad.strip()

    return texto, None
```

```python
titulo, ciudad = separar_titulo_ciudad(
    oferta["titulo_puesto"]
)

print(titulo)
print(ciudad)
```

# Segunda transformación
Actualizar el diccionario.

```python
oferta["titulo_puesto"] = titulo

oferta["ciudad"] = ciudad
```
